# Retail Sales – Exploratory Data Analysis

**Objective:** Explore sales trends, customer demographics, product performance and actionable business insights.

## Setup
Run the following cell. Install missing packages with `pip install -r ../../requirements.txt` from the repository root.

```python
import pandas as pd, numpy as np, matplotlib.pyplot as plt, seaborn as sns
sns.set_theme()

df=pd.read_csv('retail_sales_synthetic.csv',parse_dates=['Date'])
print('Shape:',df.shape); display(df.head()); display(df.dtypes); display(df.isna().sum())
display(df.describe(include='all'))
num=df.select_dtypes(include='number')
display(pd.DataFrame({'mean':num.mean(),'median':num.median(),'mode':num.mode().iloc[0],'std':num.std()}))
df['Month']=df.Date.dt.to_period('M').astype(str); df['Quarter']=df.Date.dt.to_period('Q').astype(str)
monthly=df.groupby('Month').Sales.sum()
monthly.plot(figsize=(12,4),title='Monthly sales'); plt.xticks(rotation=90); plt.ylabel('Sales'); plt.show()
print('Observation: use peaks and troughs to plan inventory and campaigns.')
quarter=df.groupby('Quarter').Sales.sum()
quarter.plot(kind='line',marker='o',figsize=(9,4),title='Quarterly sales'); plt.xticks(rotation=45); plt.show()
df['Age_Group']=pd.cut(df.Age,[17,25,35,45,55,100],labels=['18-25','26-35','36-45','46-55','56+'])
df.Age_Group.value_counts().plot(kind='bar',title='Customer age groups'); plt.show()
df.Gender.value_counts().plot(kind='pie',autopct='%1.1f%%',title='Gender breakdown'); plt.ylabel(''); plt.show()
display(df.groupby('Product').Quantity.sum().nlargest(10))
df.groupby('Category').Sales.sum().sort_values().plot(kind='barh',title='Revenue by category'); plt.show()
sns.heatmap(num.corr(),annot=True,cmap='coolwarm'); plt.title('Numerical correlation'); plt.show()
df.groupby('Category').agg(orders=('Order_ID','count'),avg_order_value=('Sales','mean')).plot(kind='scatter',x='orders',y='avg_order_value',title='Category volume vs average order value'); plt.show()
print('Recommendations: (1) allocate inventory to high-revenue categories, (2) target age groups with tailored campaigns, (3) investigate low-sales months with promotions.')
```
